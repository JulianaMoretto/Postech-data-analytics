Como conseguir isso:
 - StandardScaler → comparação estatística
-  MinMaxScaler → comparação relativa (0 a 1)

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
#divisão do modelo em teste e treino
x_train, x_test, y_train, y_test = train_test_split(x,y,random_state = 7)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
x_train
```

download.png

```python
#o minmaxscaler serve para normalizar os dados e com isso todos os valores ficam entre 0 e 1
scaler = MinMaxScaler()
#fez isso apenas para os dados de treino
scaler.fit(x_train)
```

```python
#Aplica a normalização usando os parâmetros aprendidos no treino.
#usa os min e max coletados do scaled de treino
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled
```

```python
model = LinearRegression()
model.fit(x_train_scaled, y_train)
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
MAE = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test,y_pred)
print("MAE: ",MAE)
print("R2: ",r2) #quanto mais próximo de 1, melhor esta prevendo os dados
```

```python
model_normal = LinearRegression()

model_normal.fit(x_train, y_train)
```

```python
y_pred_normal = model_normal.predict(x_test)
```

```python
MAE = mean_absolute_error(y_test, y_pred_normal)
r2 = r2_score(y_test,y_pred_normal)
print("MAE: ",MAE)
print("R2: ",r2) #quanto mais próximo de 1, melhor esta prevendo os dados
```

MAE e R² deram iguais porque a regressão linear ajusta os coeficientes de forma a compensar a escala, mantendo as previsões idênticas.
